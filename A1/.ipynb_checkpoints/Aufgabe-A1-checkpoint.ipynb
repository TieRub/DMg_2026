```python
import polars as pl
import altair as alt
alt.data_transformers.enable("vegafusion")
```

## Aufgabe A

```python
# Textdateien einlesen und DataFrames anpassen
queries = []

# Reihenfolge der einzulesenden Dateien
patharray = ['./InputsA1/iw_article.txt','./InputsA1/iw_codeReason.txt','./InputsA1/iw_customer.txt','./InputsA1/iw_payment.txt','./InputsA1/iw_return_header.txt','./InputsA1/iw_return_line.txt','./InputsA1/iw_sales.txt','./InputsA1/plz_mapping.txt','./InputsA1/productGroup.txt']

for file in patharray:
    q = pl.scan_csv(file,has_header=True,separator='\t',schema_overrides={"postcode": pl.String,"articleOnline":pl.String})\
    .drop(["articleOnline","seasonCode","modtime","deftime","article_No","owner","street","VATpercent","bill_customerNo",
           "dueDate","closedAccountDate","openAccount","paymentCode","returnType","shippingAgent","shipmentDate"],strict=False)\
    .rename({"description":"Beschreibung","unitPrice":"Preis","productGroup":"Produktgruppe","salutation":"Anrede","firstname":"Vorname","surname":"Nachname","postcode":"plz","city":"Stadt"
                ,"eMail":"EMail","colorCode":"Farbcode","colorDescription":"Farbbeschreibung","size":"Größe","newsletter":"Newsletter","birthdate":"Geburtstag","orderNo":"Bestellnummer",
             "customerNo":"Kundennummer","line_No":"Zeilennummer","vat_line_amount":"vat_zeile_amount","quantity":"Anzahl","line_amount":"zeile_amount","orderDate":"Bestelldatum","outstandingAmount":"OffenerBetrag","postingDate":"Lieferdatum",
             "dunningLevel":"Mahnstufe","returnNo":"Retourennummer","returnReason":"Retourengrund","PLZ":"plz"},strict=False)
    queries.append(q)
dataframes = pl.collect_all(queries)

##Datentypen anpassen
dataframes[0] = dataframes[0].with_columns(pl.col("Produktgruppe").cast(pl.String))
dataframes[8] = dataframes[8].with_columns(pl.col("Produktgruppe").cast(pl.String))
dataframes[2] = dataframes[2].with_columns(pl.col("Newsletter").cast(pl.Int8),pl.col("creditLimit").cast(pl.String),pl.col("riskID").cast(pl.String),pl.col("Kundennummer").str.replace_all(r"[A-Za-z]","").alias("Kundennummer").cast(pl.Int64),pl.col("Geburtstag").str.to_datetime())
dataframes[3] = dataframes[3].with_columns(pl.col("Bestellnummer").str.replace_all(r"[A-Za-z]","").alias("Bestellnummer").cast(pl.Int64),pl.col("Kundennummer").str.replace_all(r"[A-Za-z]","").alias("Kundennummer").cast(pl.Int64),pl.col("Lieferdatum").str.to_datetime(),pl.col("Mahnstufe").cast(pl.String))
dataframes[6] = dataframes[6].with_columns(pl.col("Bestellnummer").str.replace_all(r"[A-Za-z]","").alias("Bestellnummer").cast(pl.Int64),pl.col("Kundennummer").str.replace_all(r"[A-Za-z]","").alias("Kundennummer").cast(pl.Int64),pl.col("Bestelldatum").str.to_datetime()).drop("type","Lieferdatum","amount")
dataframes[7] = dataframes[7].with_columns(pl.col("plz").cast(pl.String))

##Joins return_header und return_line
dataframes[5] = dataframes[5].join(
    dataframes[4].select(["Retourennummer","Bestellnummer",]),
    on="Retourennummer",
    how="left"
).with_columns(pl.col("Retourennummer").str.replace_all(r"[A-Za-z]","").alias("Retourennummer").cast(pl.Int64),pl.col("Kundennummer").str.replace_all(r"[A-Za-z]","").alias("Kundennummer").cast(pl.Int64),pl.col("Bestellnummer").str.replace_all(r"[A-Za-z]","").alias("Bestellnummer").cast(pl.Int64)).drop("Preis","type","zeile_amount","Lieferdatum","Anzahl")

# DataFrames benannt
Artikel = dataframes[0]
Kunde = dataframes[2]
Buchung = dataframes[3]
Retoure = dataframes[5]
Bestellung = dataframes[6]
```

## Aufgabe b

```python
# 5 jüngste Kunden ausgeben
print(Kunde.sort(by = "Geburtstag",descending=True)[:5])
```

## Aufgabe c

```python
# Nur Kunden ab 1930 geboren
Kunde = Kunde.sql("DELETE FROM self WHERE Geburtstag < '1930-01-01 00:00:00'")
print(Kunde.sort(by = "Geburtstag",descending=False))
```

## Aufgabe d

```python
# Bundesland der Kunden zugeordnet
Kunde = Kunde.join(dataframes[7].select(["plz","Bundesland"]),on="plz",how="left")
print(Kunde)
```

## Aufgabe e

```python
# Eindeutige Kunden
uniqueCustomers = Kunde.unique(subset=["riskID"])
print(uniqueCustomers.count())
print(Kunde.count())
```

## Aufgabe f

```python
# Länge der Beschreibungs-Namen
Artikel = Artikel.with_columns(pl.col("Beschreibung").str.len_chars().alias("Länge"))
print(Artikel)
```

## Aufgabe h

```python
# Balkendiagramm: Verteilung der Kunden (männlich/ weiblich) je Bundesland 
group_df = uniqueCustomers.group_by(["Bundesland","Anrede"]).len().rename({"len": "Anzahl"})
group_df.plot.bar(x="Bundesland",y='Anzahl',color='Anrede')
```

## Aufgabe i

```python
# Unterteilung in 3 Kategorien der Artikelpreise
q33 = Artikel['Preis'].quantile(0.33)
q66 = Artikel['Preis'].quantile(0.66)
Artikel = Artikel.with_columns(pl.when(pl.col("Preis") <= q33).then(pl.lit("günstig")).when((pl.col("Preis") <= q66) & (pl.col("Preis") > q33))\
                               .then(pl.lit("mittel")).otherwise(pl.lit("teuer"))\
                               .alias("Preiskategorie"))

print(Artikel.group_by("Preiskategorie").len())
```

## Aufgabe j

```python
# Bestellungen mit Retouren zusammenfassen
Bestellung2010 = Bestellung.join(
    Retoure.select(["Retourennummer","Bestellnummer",]),
    on="Bestellnummer",
    how="left"
)

# Zeitraum Jahr 2010
date = pl.DataFrame({
    "Datum": ["2010-01-01 00:00:00", "2010-12-31 23:59:59"]
}).with_columns(
    pl.col("Datum").str.to_datetime("%Y-%m-%d %H:%M:%S")
)

# Bestellungen in 2010 ohne zusammenhängende Retouren filtern
Bestellung2010 = Bestellung2010.filter(pl.col("Bestelldatum").is_between(date["Datum"][0], date["Datum"][1])).sort("Retourennummer")
Bestellung2010 = Bestellung2010.filter(pl.col("Retourennummer").is_null())

# Bestelldatum in i-ten Bestellmonat konvertieren
Bestellung2010 = Bestellung2010.with_columns(pl.col("Bestelldatum").dt.month().alias("Bestellmonat"))

# Gesamte Bestellwertsummen der Monate ermitteln 
SummeBestellung2010 = Bestellung2010.group_by("Bestellmonat").agg(pl.col("zeile_amount").sum().alias("Summe"))
SummeBestellung2010 = SummeBestellung2010.extend(pl.DataFrame({"Bestellmonat": [1,2,3,4], "Summe":[0.0,0.0,0.0,0.0]}, schema=[("Bestellmonat", pl.Int8), ("Summe", pl.Float64)])).sort("Bestellmonat")

# Balkendiagramm: Summe der Bestellwerte pro Monat
SummeBestellung2010.plot.bar(x="Bestellmonat", y="Summe")
```

## Aufgabe k

```python
# Mahnstufen zur Bestellung ergänzen
MahnstufeProduktgruppe = Bestellung.join(Buchung.select(["Bestellnummer", "Mahnstufe"]), on="Bestellnummer", how="left").join(Artikel.select(["IWAN", "Produktgruppe"]), on="IWAN", how="left")
MahnstufeProduktgruppe = MahnstufeProduktgruppe.with_columns(pl.col("Mahnstufe").cast(pl.Int8))

# Durchschnittliche Mahnstufe ermitteln
DurchschnittMahnstufeProduktgruppe = MahnstufeProduktgruppe.group_by("Produktgruppe").agg(pl.col("Mahnstufe").mean().alias("MahnstufeDurchschnitt")).sort(by="MahnstufeDurchschnitt", descending=True)

# Zeige die 10 Produktgruppen mit den höchsten durchschnittlichen Mahnstufen
print(DurchschnittMahnstufeProduktgruppe[:10])
```

## Aufgabe l

```python
# Durchschnittliche Anzahl an Bestellpositionen ermitteln
AnzahlBestellposition = Bestellung.group_by("Bestellnummer").agg(pl.col("Zeilennummer").max().alias("MaxBestellposition"))

print("Durchschnittliche Anzahl an Bestellpositionen: ", AnzahlBestellposition["MaxBestellposition"].mean())

# Durchschnittlichen Bestellwert ermitteln
SummeBestellnummer = Bestellung.group_by("Bestellnummer").agg(pl.col("zeile_amount").sum().alias("SummeBestellung"))

print("Durchschnittlicher Bestellwert: ", SummeBestellnummer["SummeBestellung"].mean())
```

## Aufgabe m

```python
# Anzahl der Bestellpositionen ermitteln
AnzahlBestellposition = Bestellung.group_by(["Bestellnummer", "Kundennummer"]).agg(pl.col("Zeilennummer").max().alias("MaxBestellposition"))

BestellungRetoure = AnzahlBestellposition.join(
    Retoure.select(["Retourennummer","Bestellnummer",]),
    on="Bestellnummer",
    how="left"
).join(uniqueCustomers.select(["Kundennummer", "Bundesland"]),
       on="Kundennummer",
       how="left"
      )

# Anzahl der Bestell- und Retourenpositionen pro Bundesland
BestellRetourePositionBundesland = BestellungRetoure.group_by("Bundesland").agg(pl.col("MaxBestellposition").sum().alias("AnzahlBestellposition"), pl.col("Retourennummer").n_unique().alias("AnzahlRetourenposition"))

# Retourenquote pro Bundesland
BestellRetourePositionBundesland = BestellRetourePositionBundesland.with_columns((pl.col("AnzahlRetourenposition") / pl.col("AnzahlBestellposition")).alias("Retourenquote"))

print(BestellRetourePositionBundesland)
```
